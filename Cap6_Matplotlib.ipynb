```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## Gráficos de linhas (Plot)

Gráficos de linhas -> Plot

**Para variações ao longo do tempo**

```python
# eixo x
x = np.array([0, 1, 2, 3, 4])

# eixo y
y = x*2

# legendas nos eixos x e y
plt.xlabel("Valores x")
plt.ylabel("Valores y")

# titulo do gráfico
plt.title("Gráfico de linhas")

# plotando o grafico
plt.plot(x, y)
```

### Formatando o Gráfico

Passando o format(fmt) no plot

```python
plt.plot(x, y, '*:r')
```

```python
plt.plot(x, y, '*:r', linewidth='3', markersize=20)
```

### Plotandos dois Gráficos no mesmo Plano Cartesiano

```python
y2 = x**2

plt.plot(x, y, '*:r', x, y2, 's--b', linewidth='2', markersize=10)
plt.show()
```

### Usando conceito de SubPlots (Dashboard)

Quando se cria um subplot, deve se passar quantas linhas e quantas colunas tem a grade subplot. Alem de falar qual a posicao do grafico na grade

```python

plt.subplot(1, 2, 1)
plt.plot(x, y, '*:r', linewidth='2', markersize=10)

plt.subplot(1, 2, 2)
plt.plot(x, y2, 's--b', linewidth='2', markersize=10)
plt.show()
```

## Gráfico de Dispeção (Scatter Plot)

**Para saber a dispeção(distancia) dos dados**

```python
ds = pd.read_csv('paises.csv', delimiter=';')

ds.head()
```

```python
print(ds.columns)
```

```python
# Quao dispersas estao as rendas per capita dos 6 maiores paises do mundo?

# Pegando apenas os 6 maiores paises do planeta
maiores = ds.nlargest(6, 'Area (sq. mi.)')

maiores
```

```python
# Tracando o grafico com as rendas per capitas desses paises
plt.scatter(maiores['Country'], maiores['GDP ($ per capita)'])
plt.show()
```

```python
# Adicionando uma terceira dimensao
# O parametro S permite adicionar uma terceira dimensao neste grafico
plt.scatter(maiores['Country'],
            maiores['GDP ($ per capita)'],
            s=maiores['Area (sq. mi.)']/10000)
plt.show()
```

## Gráfico em Barras (Bar Plot)

**Compara a quantidade de dados em cada coisa (dados quantidados)**

```python
# Qual a diferenca das 5 maiores rendas per capita do dataset?

maiores = ds.nlargest(5, 'GDP ($ per capita)')
maiores
```

```python
plt.bar(maiores['Country'], maiores['GDP ($ per capita)'])
plt.show()
```

## Gráfico em Torta/Pizza (Pie Plot)

**Mostra as proporções dos dados**

```python
# Pegando os paises que nao possuem costa maritma
paises_semCosta = ds[ds['Coastline (coast/area ratio)'] == 0]
paises_semCosta
```

```python
# Calculando a quantidade e paises sem costa e quantidade com costa
quant_semCosta = len(paises_semCosta)
quant_comCosta = len(ds) - quant_semCosta

print(quant_semCosta, quant_comCosta)
```

```python
plt.pie(x=[quant_semCosta, quant_comCosta],
        labels=['% Paises sem costa',
                '% Paises com costa'],
        autopct='%1.2f%%')

plt.show()
```
