1.   Crie um programa que **leia seu nome completo** e mostre:

- Seu nome com **todas as letras maiúsculas**

- Seu nome com **todas as letras minúsculas**

- **Quantas letras** ao todo tem seu nome

- E como seria se **trocássemos** seu último nome para “do Inatel”



```python
nome = input("Insira seu nome completo: ")

print("\n" + nome.upper())

print("\n" + nome.lower() + "\n")

# Aqui será contado os espaços também
# tentar criar algo para desconsiderar (pode trim)
print(len(nome))

# tentar criar uma logica para pegar o ultim nome
# pode usar o split
print("\n" + nome.replace("Siécola", "do Inatel"))
```


2.    Mostre a **tabuada** de um número que o usuário escolher **dentro de um intervalo** específico também escolhido por ele




```python
comeco = int(input("Insira o começo do intervalo: "))
fim = int(input("Insira o final do intervalo: "))

num = int(input("\nInsira o número que deseja ter a tabuada: "))

print("\nTABUADA")

for i in range(comeco, fim + 1):
  print("{} x {} = {}".format(num, i, num * i))
```

3. Faça um programa que **leia o sexo** de uma pessoa e **diga** se ela é homem (caso seja digitado M) ou mulher (caso seja digitado F). Caso seja digitado algo **inválido**, **continue perguntando** até que o usuário entre com um sexo válido

```python
sexo = input("Qual o seu sexo?(M = Homem ou F = Mulher)")

while sexo != "M" and sexo != "F":
  print("Sexo inválido")
  sexo = input("\nQual o seu sexo?(M = Homem ou F = Mulher)")

if sexo == "M":
  print("Você é homem")
elif sexo == "F":
  print("Você é mulher")
```

## Exercícios (Parte 2)

4. Desenvolva um script que **pergunte a distância** de uma viagem em
Km. **Calcule o preço da passagem**, cobrando R\$0.50 por Km para viagens até 200Km e R\$0.45 para viagens mais longas


```python
distancia = int(input("Qual a distancia da sua viagem? "))

if distancia <= 200:
  preco = distancia * 0.50
elif distancia > 200:
  preco = distancia * 0.45

print("Sua passagem custará : R${}".format(preco))
```

5. Faça um programa que **leia um número entre 1000 e 9999** e mostre na tela:

- Número da **unidade**

- Número da **dezena**

- Número da **centena**

- E número do **milhar**

```python
num = input("Insira um número que esteja entre 1000 e 9999: ")

print("Unidade:", num[3:])
print("Dezena:", num[2:3])
print("Centena:", num[1:2])
print("Milhar:", num[0:1])
```

6. Peça ao usuário para entrar com um número decimal. Em seguida, aplique e mostre o resultado:

- Raiz quadrada deste número

- Função teto

- Função chão

- Sua parte inteira

```python
import math

num = float(input("Insira um  numero decimal: "))

print("\nRaiz quadrada: ", math.sqrt(num))

print("\nFunção teto: ", math.ceil(num))

print("\nFunção chão: ", math.floor(num))

print("\nParte inteira: ", math.trunc(num))
```

7. Faça um programa que **leia uma palavra**. Esse programa deve percorrer a palavra e **imprimir cada letra em maiúsculo**, uma letra por linha. No final, deverá informar **quantas vogais** a palavra tem e **se a letra ‘A’ está presente** nela

```python
palavra = input("Insira uma palavra: ")

print("\nLetra por letra em maiúsculo:(Opção 1) ")
palavra = palavra.upper()
for i in range(len(palavra)):
  print(palavra[i])
# Outra opcao:
print("\nLetra por letra em maiúsculo:(Opção 2) ")
for i in palavra:
  print(i)

print("\nQuantidade de vogais:(Opção 1) ")
quant = 0
for i in palavra:
  if i == "A":
      quant += 1
  if i == "E":
      quant += 1
  if i == "I":
      quant += 1
  if i == "O":
      quant += 1
  if i == "U":
      quant += 1
print(quant)

# Opcao gerada pelo colab ( mas deu para entender :) )
print("\nQuantidade de vogais:(Opção 2) ")
quant = 0
for i in palavra:
  if i in "AEIOU":
    quant += 1
print(quant)

print("\nA letra A está presente na palavra? ")
if "A" in palavra:
  print("Sim, a letra 'A' está presente")
else:
  print("Não, a letra 'A' não está presente")
```

8. Crie um programa que **leia dois números** calcule e mostre na tela:

- **Adição**

- **Subtração**

- **Multiplicação**

- **Divisão**

- **Resto da divisão**

- **Potência** (primeiro número elevado ao segundo)

```python
num1 = int(input("Insira o primeiro número: "))
num2 = int(input("Insira o segundo número: "))

print("\nAdição:", num1 + num2)
print("\nSubtração:", num1 - num2)
print("\nMultiplicação:", num1 * num2)
print("\nDivisão:", num1 / num2)
print("\nResto da divisão:", num1 % num2)
print("\nPotência:", num1 ** num2)
```
