Baseado no dataset paises.csv fornecido, crie códigos em Python que

```python
import numpy as np
```

```python
dataset = np.loadtxt('paises.csv', delimiter = ';', dtype = str, encoding='utf-8')
dataset = np.char.strip(dataset) #Retira espaços em branco
```

## Questão 1
Faça um slicing no dataset para mostrar apenas o País (Country), Região
(Region), População (Population) e Area (Area (sq. mi.)) dos países contidos
nele;

```python
# Descobrindo quais colunas existem
print(dataset[0,:])

# Mostrando tudo junto
print("\n-> Todas as colunas que pediu juntas: ")
print(dataset[1:,:4]) # Lembrando que o 4 não inclui e o 1 sim

# Mostrando coluna por coluna
print("\n-> Coluna 'Country': ")
print(dataset[1:,0])

print("\n-> Coluna 'Region': ")
print(dataset[1:,1])

print("\n-> Coluna 'Population': ")
print(dataset[1:,2])

print("\n-> Coluna 'Area sq. mi.': ")
print(dataset[1:,3])
```

## Questão 2
Conte e em seguida mostre quais são as diferentes Regiões do planeta segundo
este dataset;


```python
# Contando a quantidade de elementos na lista criada pelo unique
print("-> Quantidade de regiões diferentes existentes:")
print(len(np.unique(dataset[:,1])))

# Mostrando quais são
print("\n-> São as seguintes regiões: ")
print(np.unique(dataset[:,1]))

# Mostrando a quantidade de cada uma
print("")
regioes, quantidade = np.unique(dataset[:,1],return_counts=True)
for i in range(len(regioes)):
  print(f"-> A região {regioes[i]} aparece {quantidade[i]} vezes")
```

## Questão 3
Mostre qual a taxa média de alfabetização (Literacy (%)) do planeta segundo
este dataset;

```python
# Calculando a media das linhas da coluna Literancy
media = np.mean(dataset[1:,9].astype(float)) # Tirando linha 1 que sao os nomes das colunas e passado para float os valores

print(f"A média de alfabetizacao é {media:.2f}%")
```

## Questão 4
Conte quantos países são da América do Norte (NORTHERN AMERICA)
segundo este dataset;


```python
# Se ACHAR a palavra da coluna regiao vai ser != -1
cond = np.char.find(dataset[:,1],"NORTHERN AMERICA") != -1

# Conta quantos "NORTHERN AMERICA" tera na lista criada que contem apenas Northen america
print("A quantidade de paises que são da America do Norte é: ")
print(len(dataset[:,1][cond]))

```

## Questão 5
Encontre qual país da América do Sul e Caribe (LATIN AMER. & CARIB)
possui a maior renda per capita (GDP ($ per capita));

```python
# Tive muita dificuldade...
# Puxando as colunas pulando o cabeçalho (linha 0) em TODAS
paises = dataset[1:, 0]
regioes = dataset[1:, 1]
renda = dataset[1:, 8].astype(float)

# Criando a máscara de condição para a região desejada
condicional = np.char.find(regioes, "LATIN AMER. & CARIB") != -1

# Filtrando os países e rendas
paises_regiao_desejada = paises[condicional]
rendas_regiao_desejada = renda[condicional]

# Encontrando o índice do maior valor e o país correspondente
indice = np.argmax(rendas_regiao_desejada)
pais_maior_renda = paises_regiao_desejada[indice]

print(f"A maior renda per capita da América do Sul e Caribe pertence ao país:\n{pais_maior_renda}")
```
