2. Crie dois NumPy Arrays unidimensionais: um de números pares de 0 à 51 e outro também de números pares de 100 até 50. Em seguida, os concatene e mostre os resultados ordenados

```python
arr1 = np.arange(0, 52, 2)
arr2 = np.arange(100, 50, -2)

print("arr1: ")
print(arr1)

print("\narr2: ")
print(arr2)
```

3. Mini Campo Minado

- Crie um NumPy Array 2x2 formado apenas por 0’s

- Em seguida, adicione um número 1 em uma posição aleatória desta matriz;

- Faça uma entrada de dados para solicitar o usuário que faça uma jogada (selecione uma posição da matriz)

- I. Se ele selecionar todas as posições em que o número 1 não se encontra, mostre a mensagem “Congratulations! You beat the game!:)”

- II. Senão, se dentro das 3 primeiras jogadas ele achar o número 1, mostre a mensagem “Game Over!:( Try Again!”

```python
# Gerando matriz 2x2 todos elementos 0's
mtz = np.zeros((2,2))

# Sorteando uma linha e coluna
linha = np.random.randint(0, 2)
coluna = np.random.randint(0, 2)

# Colocando um 1 na posição sorteada
mtz[linha, coluna] = 1

print(mtz) # Para testar a celula abaixo
```

```python
# Contador auxiliar para saber se a pessoa nao errou nada
cont = 0

for i in range(3):
  linha = int(input("\nDigite a linha: "))
  coluna = int(input("Digite a coluna: "))

  if mtz[linha, coluna] == 1:
    print("\nGame Over!:( Try Again!")
    break
  else:
    # Somando 1 para cada acerto
    cont+=1

# Se acertar 3 sem acertar o 1 mostra mensagem
if cont == 3:
  print("\nCongratulations! You beat the game!:)")
```

4. Crie uma matriz de tamanho qualquer. Extraia seu número de linhas e colunas, multiplique-os, e diga se esta matriz poderia se tornar um vetor unidimensional com número par ou ímpar de elementos

```python
linha = np.random.randint(0, 6)
coluna = np.random.randint(0, 6)

arr = np.random.randint(1, 51, linha*coluna)

mtz = arr.reshape(linha, coluna)

print(mtz)
```

```python
quant_col = mtz.shape[1]
quant_lin = mtz.shape[0]

print("Quantidade de colunas: ", quant_col)
print("Quantidade de linhas: ", quant_lin)
print("Quantidade de elementos: ", quant_col*quant_lin)

if quant_col*quant_lin % 2 == 0:
  print("\nPode virar vetor unidimensional com numero PAR de elementos")
else:
  print("\nPode virar vetor unidimensional com numero IMPAR de elementos")
```

5. Crie uma matriz de tamanho 4x4 formada por números aleatórios inteiros entre 1 e 50 (use seed = 10 antes)

- Mostre o resultado da média de cada linha e cada coluna da matriz gerada

- Apresente o maior valor das médias das linhas e também das colunas

- Mostre a quantidade de aparições de cada um dos números gerados na
matriz. Em seguida, mostre apenas os números que aparecem 2 vezes

```python
np.random.seed(10)

arr = np.random.randint(1, 51, 16)
mtz = arr.reshape(4, 4)

print(mtz)

# Media de cada linha e maior valor
print("\nMedia de cada linha: ")
print(mtz.sum(axis=1)/4)
print("Maior valor: ")
print((mtz.sum(axis=1)/4).max())

# Media de cada coluna e maior valor
print("\nMedia de cada coluna: ")
print(mtz.sum(axis=0)/4)
print("Maior valor: ")
print((mtz.sum(axis=0)/4).max())

# Quantidade de apariçõe de cada numero
print("\nQuantidade de aparições de cada numero: ")
aparicoes = np.unique(mtz, return_counts = True)
print(aparicoes)

# Aparecem 2 vezes
print("\nAparecem 2 vezes: ")
for i in range(len(np.unique(mtz))):
  if aparicoes[1][i] == 2:
    print(aparicoes[0][i])
```

```python
dataset = np.loadtxt('space.csv', delimiter = ';', dtype = str)

print(dataset)
print("\n-------------------------------------------------------\n")

# Extraindo as colunas do dataset
print(dataset[0,:])
```

1. Apresente a porcentagem de missões que deram certo

```python
porcentagem = np.sum(dataset[1:,-1] == 'Success')/len(dataset[1:,-1])

print(f"Porcentagem de missoes que deram certo: {porcentagem*100:.2f}%")
```

2. Qual a media de gastos de uma missão especial se baseando em missões que possuam valores disponíveis (> 0)?

```python
gastos = np.array(dataset[1:, -2]).astype(float)

print(f"Media de gastos: {np.mean(gastos[gastos > 0]):.2f}")
```

3. Encontre quantas missões espaciais neste Dataset foram realizadas pelos Estados Unidos (EUA)

```python
print("Quantidade de missoes realizadas pelos EUA:")


print(np.sum(np.char.find(np.char.upper(dataset[1:,2]), "US") !=-1 ))
```

4. Encontre qual foi a missão mais cara realizada pela empresas “SpaceX”

```python
cond = (dataset[1:, 1] == 'SpaceX') & (dataset[1:, 6] != '')

missoes = dataset[1:, 4][cond]
gastos = dataset[1:, 6][cond].astype(float)

print(f"Missao mais cara: {missoes[np.argmax(gastos)]}")
print(f"Valor da missao: {gastos[np.argmax(gastos)]:.2f}")
```
