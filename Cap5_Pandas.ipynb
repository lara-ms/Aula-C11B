# Pandas

```python
import numpy as np
import pandas as pd
```

## Pandas Series (1-D)

### Como preencher uma Series

```python
# Lista de labels
labels = ['Tiago', 'Mateus', 'Bruna', 'Julia']

# Lista de valores
valores = [23, 25, 27, 22]
```

### Criando a Series

```python
se1 = pd.Series(index=labels, data=valores)

print(se1)

print(type(se1))
```

### Acessando os valores da Series

```python
print(se1['Mateus'])

print(se1[['Mateus', 'Julia']])
```

### Operações de Series

```python
s1 = pd.Series({'v1' : 10,'v2' : 45,'v3' : 70})
s2 = pd.Series({'v1' : 10,'v4' : 50,'v3' : 70})

print(s1)
print("\n")

print(s2)
print('\n')

print(s1 + s2)
print('\n')

print(s1.add(s2, fill_value=0))
```

## Pandas Dataframe (2-D)

### Como preencher DataFrame

```python
# Lista de labels (COLUNAS)
colunas = ['W', 'X', 'Y', 'Z']

# Lista de labels (LINHAS)
linhas = ['A', 'B', 'C', 'D', 'E']

# Lista de valores
np.random.seed(10)
valores = np.random.randint(1, 50, [5, 4])
print(valores)
```

### Criando DataFrame

```python
df = pd.DataFrame(columns = colunas,
                  index = linhas,
                  data = valores)

print(df)
```

### Manipulando DataFrame

```python
# Puxando uma unica coluna
print(df['X'])
print(type(df['X']))
print("\n")

# Puxando mais de uma coluna
print(df[['Y', 'Z']])
print(type(df[['Y', 'Z']]))
print("\n")

# Puxando uma unica celula [COLUNA][LINHA]
print(df['Y']['C'])
print('\n')

# Puxando multiplas colunas
print(df[['W', 'X', 'Z']])
print('\n')

```

### Slicing de dados no dataframe com LOC (Labels) e ILOC (Indices Numericos)


LOC

```python
# Puxando uma unica linha
print(df.loc['C', ['W', 'X', 'Y', 'Z']])
print("\n")
print(df.loc['C'])
print('\n')

# Puxando mais de uma linha
print(df.loc[['B', 'E']])
print('\n')

# Puxando o quadrado do canto inferior direito
print(df.loc[['D','E'], ['Y', 'Z']])
```

ILOC

```python
# Puxando uma unica linha
print(df.iloc[2, :]) # -> df.iloc[2]
print("\n")

# Puxando mais de uma linha
print(df.iloc[[1,4]])
print('\n')

# Puxando o quadrado do canto inferior direito
print(df.iloc[[3,4], [2,3]])
```
