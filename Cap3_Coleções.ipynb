## 1. Tuplas

- É a **forma mais simples de coleção**.

- A organização dos dados é feita dentro de **()**




```python
nomes = ('Goku', 'Vegeta', 'Trunks', 'Gohan')

print(nomes)

print (type(nomes))
```

- É uma coleção **IMUTÁVEL**

```python
# Tentando adicionar
# nomes[4] = 'Piccolo'

# Tentando mudar
# nomes[3] = 'Majin Boo'
```

- Permite **SLICING**

Slicing de dados = fatiar os dados, pegar apenas o que deseja no seu dado  


```python
print(nomes)

print(nomes[1])

print(nomes[1:3])

print(nomes[2:])

print(nomes[-2])
```

## 2. Listas

- A organização dos dados é feita dentro de **[]**

```python
nomes = ['Goku', 'Vegeta', 'Trunks', 'Gohan']

print(nomes)

print(type(nomes))
```

- É uma coleção **MUTÁVEL**

- Permite **SLICING**


- **CRUD de dados**

CREATE

```python
nomes.append('Piccolo') # add no final
nomes.insert(3, 'Majin Boo') # ele nao apaga apenas empurra

print(nomes)
```

READ

```python
print(nomes[3])

print(nomes[1:3])

print(nomes[2:])

print(nomes[-3])
```

UPDATE

```python
nomes[3] = 'Tenshin Han'

print(nomes)
```

DELETE

```python
del nomes[1]

print(nomes)

nomes.remove('Trunks')

print(nomes)
```

## 3. Conjuntos

- É uma coleção que **não guarda ordem** e **não admite elementos duplicados**

- A organização dos dados é feita dentro de **{}**

## 4. Dicionários


- É uma coleção **MUTÁVEL**

- A organização feita por **CHAVE:VALOR**

```python
pessoa = {
    'nome':'Goku',
    'idade':25 ,
    'sexo': 'M'
    }

print(pessoa)

print(type(pessoa))
```

CREATE

```python
pessoa['Desenho'] = 'DBZ'

print(pessoa)
```

READ

```python
print(pessoa['nome'])

print(pessoa['idade'])
```

UPDATE

```python
pessoa['Desenho'] = 'DBGT'

print(pessoa)
```

DELETE

```python
del pessoa['sexo']

print(pessoa)
```
