## Numpy Arrays

Para usar o **NdArray** pode criar **tuplas e listas** para usar no numpy.

O NumPy Array guarda apenas elementos do mesmo tipo
• Isso fornece maior desempenho nas operações
• Menor uso de memóri

```python
# Importando o numpy
import numpy as np
```

```python
# Criando um numpy array
arr = np.array([21, 20, 49, 17])

print(arr)

print(type(arr))
```

Propriedades do Array

```python

print(arr.size)  # numero de elementos

print(arr.ndim)  # dimensao

print(arr.shape) # formato
```

```python
mtz = np.array([[10, 20, 30],[11, 21, 31],[12, 22, 32],[13,  23, 33]])

print(mtz)

print(type(mtz))

```

Propriedades de Matriz (Array de 2 dimensões)  

```python

print(mtz.size)  # numero de elementos

print(mtz.ndim)  # dimensao

print(mtz.shape) # formato
```

## Funções pré prontas para estruturas de Numpy Arrays

```python
# Ones
mtz = np.ones([5,5])
print(mtz)
```

```python
# Zeros
arr = np.zeros(10)
print(arr)
```

```python
# Transformando uum Array em Matriz
print(arr.reshape(5,2))
```

```python
# Arange
mtz = np.arange(2, 21, 2).reshape(2, 5)

print(mtz)
```

## Operações entre Numpy Arrays

```python
arr1 = np.array([10, 20, 30, 40])
arr2 = np.array([1, 2, 3, 4])
arr3 = arr1 + arr2

print(arr3)
print(arr1 - arr2)
print(arr1 * arr2)
print(arr1 / arr2)

# BROADCASTING - qnd um ESCALAR faz uma operacao com um ARRAY
print(5 * arr3)

# Concatenar Arrays
arr3 = np.concatenate([arr1, arr2])
print(arr3)
```

```python
# Estruturando uma matriz com contas
mtz = np.arange(10, 96, 5).reshape(3, 6)

print(mtz)

# Extraindo a soma da coluna 1
print(mtz.sum(axis=0)[0]) # eixo 0 = coluna

print(mtz.sum(axis=1)[1]) # eixo 1 = linha
```

Numeros aleatorios em Numpy

```python
arr = np.random.randint(10)
print(arr)
```

```python
np.random.seed(5) # platando a semente aleatoria
arr = np.random.randint(1, 10, 10)
print(arr)
```

```python
# Extraindo elementos unicos
print(np.unique(arr))

# Contando elemelementos unicos
print(np.unique(arr, return_counts=True))
```

## Slicing no NumPy

```python
import numpy as np
```

```python
# Plantando a mesma semente aleatoria
np.random.seed(10)

# Estruturando a matmatriz 3x3
mtz = np.random.randint(1, 99, 9).reshape(3,3)

print(mtz)
```

```python
# Extraindo apenas a segunda linha da matriz
print(mtz[1])
print("--------------------------")

# Extraindo apenas a terceira coluna da matriz
print(mtz[:,2])
print("--------------------------")

# Extraindo a matriz 2x2 no canto inferior direito da matriz
print(mtz[1:,1:])
```

## Condicionais no NumPy

voce mata as estruturas de repeticao de programacao   

```python
print(mtz)
print("--------------------------")

# Mostre apenas os elementos menores que 70
print(mtz < 70)

print(mtz[mtz < 70])
print("--------------------------")

# Retorne apenas os numeros pares
print(mtz[mtz % 2 == 0])
```

## Analise de Padroes textuais com NumPy

```python
arr = np.array(['Inatel', 'Casa Viva', 'ICC',
                'CDG', 'eHealth', 'CSILab',
                'RobotBulls', 'ProdLab', 'CRA',
                'CRR', 'Ave'])

print(arr)
```

### Char

```python
# Buscando qual texto aceita um padrao informado
print(np.char.find(arr, 'a'))
print("--------------------------")

print(np.char.find(arr, 'a') >= 0)
print("--------------------------")

print(arr[np.char.find(arr, 'a') >= 0])
print("--------------------------")

# Padronizando para maiusculo (upper) ou minusculo (lower)
arr = np.char.upper(arr)
print(arr[np.char.find(arr, 'A') >= 0])
```

## Importando Dataset no Numpy

No colab deve sempre colocar dnv o dataset em arquivos pq sempre vai apagar

```python
dataset = np.loadtxt('space.csv', delimiter = ';', dtype = str)

print(dataset)
```

```python
# Extraindo as colunas do dataset
print(dataset[0,:])
print("-------------------------------------------------------")

# Extraindo o nome das empresas
print(dataset[1:,1]) # Tirando o nome da coluna com '1:'

# Extraindo os nomes unicos, para nao aparecer nomes repetidos
print(np.unique(dataset[1:,1]))
print("-------------------------------------------------------")

print(np.unique(dataset[1:,1], return_counts=True))
```
