## Questão 2
Para esta questão, faça o que se pede em ordem:
- Crie dois NumPy Arrays 1-D com 4 nomes de pessoas cada;

- Em seguida, concatene-os em um só Array;

- Transforme o Array final em um Array 2-D com mais colunas
do que linhas;

- Por fim, ordene os nomes do Array 2-D em ordem
decrescente;

````python
import numpy as np

arr1 = np.array(['Lara', 'Pedro', 'Gigi', 'Mari'])
print(f"Primeiro array : {arr1}")
arr2 = np.array(['Duda', 'Thais', 'Nath', 'Sara'])
print(f"\nSegundo array : {arr2}")

# concatenando os arrays
arr_conca = np.concatenate((arr1, arr2))
print(f"\nArrays concatenados : {arr_conca}")

# tranformando me matriz
mtz = arr_conca.reshape(2,4)
print(f"\nMatriz gerada com 4 colunas e 2 linhas :")
print(mtz)

# ordedando as linhas
mtz_decr = np.sort(mtz, axis=1)[:, ::-1]
print(f"\nMatriz em ordem descrescente :")
print(mtz_decr)

# ordedando mas usando o arr depois passando para matriz
mtz_decr = np.sort(arr_conca)[::-1].reshape(2, 4)
print(f"\nMatriz em ordem decrescente: (usando base o arr_conca)")
print(mtz_decr)
````

## Questão 3
Os valores de cores RGBA são uma extensão dos valores de cores
RGB com um canal alfa que especifica a opacidade de uma cor. Um
valor de cor RGBA é especificado com:
rgba (vermelho, verde, azul, alfa). O parâmetro alfa é um número
entre 0,0 (totalmente transparente) e 1,0 (totalmente opaco).

Baseado nestas informações e na coleção abaixo, crie códigos e
imprima os resultados que respondam às seguintes perguntas:


```
colors = [
{"color": "black", "type": "primary", "code": {"rgba": [255,255,255,1],"hex": "#000"}},
{"color": "green", "type": "secondary", "code": {"rgba": [0,255,0,0.1],"hex": "#0F0"}},
{"color": "yellow", "type": "primary","code": {"rgba": [255,255,0,0.7],"hex": "#FF0"}},
{"color": "blue", "type": "primary","code": {"rgba": [0,0,255,1],"hex": "#00F"}}
]
```

- Mostre apenas o nome das cores que são primárias;

- Mostre apenas os códigos hexadecimais das cores que possuem tom de
azul máximo (255);

- A partir da coleção colors, extraia seus valores para criar um NumPy
Array 1-D formado apenas pelo nome e código hexadecimal de cada
cor;
Ex: “black”, ”#000”, “green”, “#0F0”...

- Transforme o Array 1-D em um Array 2-D no seguinte padrão:

...

green #0F0

yellow #FF0

...

- Troque o nome de cada cor no Array 2-D para seu respectivo nome em
Português;

````python
colors = [
{"color": "black", "type": "primary", "code": {"rgba": [255,255,255,1],"hex": "#000"}},
{"color": "green", "type": "secondary", "code": {"rgba": [0,255,0,0.1],"hex": "#0F0"}},
{"color": "yellow", "type": "primary","code": {"rgba": [255,255,0,0.7],"hex": "#FF0"}},
{"color": "blue", "type": "primary","code": {"rgba": [0,0,255,1],"hex": "#00F"}}
]
````

````python
# achando os que tem primary como type
primarias = [c['color'] for c in colors if c['type'] == 'primary']
print(f"Cores primárias: {primarias}")

# achando os que tem 255 na 3 posicao
azul_max = [c['code']['hex'] for c in colors if c['code']['rgba'][2] == 255]
print(f"\nCódigos hex com azul máximo: {azul_max}")

# criando um array com os nomes e seus hexadecimais
lis = []
for c in colors:
    lis.append(c['color'])
    lis.append(c['code']['hex'])
arr_1d = np.array(lis)
print(f"\nArray :\n{arr_1d}")

# transformando em matriz
mtz = arr_1d.reshape(-1, 2)
print(f"\nMatriz :")
print(mtz)

# mudando o nome das cores para portugues
mtz[:, 0] = ['preto', 'verde', 'amarelo', 'azul']
print(f"\nMatriz com os nomes em portugues :")
print(mtz)
````
